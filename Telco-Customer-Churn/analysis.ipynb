```python
# Data Cleaning
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df = df.drop_duplicates()
df
```

```python
#Feature Engineering
```

```python
#Tenure Group
df["TenureGroup"] = pd.cut(df["tenure"],
                           bins = [-1,12,24,36,48,72],
                        labels = ["0-12","13-24","25-36","37-48","49-72"])
```

```python
df["TenureGroup"]
```

```python
#Monthly Charge Category
df["ChargeCategory"] = pd.cut(
    df["MonthlyCharges"],
    bins=[0, 35, 70, 120],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)
df["ChargeCategory"]
```

```python
#Churn Numeric

df["ChurnFlag"]  = df["Churn"].map({"Yes" : 1, "No" : 0})
df["ChurnFlag"]
```

```python
# Exploratory Data Analysis (EDA)
```

```python
# Total Customers
total_customers = df["customerID"].nunique()
total_customers
```

```python
#churn Distribution
print(df["Churn"].value_counts())
```

```python
#Churn Rate
churn_rate = df["ChurnFlag"].mean() * 100
churn_rate 
```

```python
#Contract-wise Churn
contract_churn = df.groupby("Contract")["ChurnFlag"].mean() * 100
contract_churn
```

```python
#Average Monthly Charges by Churn
print(df.groupby("Churn")["MonthlyCharges"].mean())
```

```python
#Revenue Analysis
```

```python
#Churned Customer Revenue

churned = df[df["Churn"] == "Yes"]
revenue_lost = churned["MonthlyCharges"].sum()
revenue_lost
```

```python
#Average Revenue per Customer

avg_revenue = df["MonthlyCharges"].mean()
avg_revenue
```

```python
#Contract-wise Revenue Risk

revenue_risk = df.groupby("Contract").agg(
    Total_Customers=("customerID", "count"),
    Churned_Customers=("ChurnFlag", "sum"),
    Revenue_At_Risk=("MonthlyCharges", "sum")
)

print(revenue_risk)
```

```python
# Contract-wise Churn
contract_churn = df.groupby("Contract")["ChurnFlag"].mean() * 100
contract_churn
```

```python
# Average Monthly Charges by Churn
df.groupby("Churn")["MonthlyCharges"].mean()
```

```python
#Internet Service-wise Churn
internet_churn = df.groupby("InternetService")["ChurnFlag"].mean() * 100
internet_churn
```

```python
# Payment Method-wise Churn
payment_churn = df.groupby("PaymentMethod")["ChurnFlag"].mean() * 100
payment_churn
```

```python
#Tenure Group-wise Churn
tenure_churn = df.groupby("TenureGroup", observed=True)["ChurnFlag"].mean() * 100
tenure_churn
```

```python
# Charge Category-wise Churn
charge_churn = df.groupby("ChargeCategory", observed=True)["ChurnFlag"].mean() * 100
charge_churn
```

```python
# Churned Customers by Contract
churned_contract = df[df["Churn"] == "Yes"]["Contract"].value_counts()
churned_contract
```

```python
# NUMPY STATISTICAL ANALYSIS

# Mean
print("Mean Monthly Charges:", np.mean(df["MonthlyCharges"]))
print("Mean Tenure:", np.mean(df["tenure"]))
```

```python
#median
print("Median Monthly Charges:", np.median(df["MonthlyCharges"]))
print("Median Tenure:", np.median(df["tenure"]))
```

```python
# standard deviation
print("Monthly Charges Standard Deviation:", np.std(df["MonthlyCharges"]))
print("Tenure Standard Deviation:", np.std(df["tenure"]))
```

```python
# variance
print("Monthly Charges Variance:", np.var(df["MonthlyCharges"]))
print("Tenure Variance:", np.var(df["tenure"]))
```

```python
#percentiles
print("Monthly Charges Percentiles:", np.percentile(df["MonthlyCharges"], [25, 50, 75]))
print("Tenure Percentiles:", np.percentile(df["tenure"], [25, 50, 75]))
```

```python
# Churned vs Retained Customer Analysis
```

```python
#separate customer analysis

churned_customers = df[df["Churn"] == "Yes"]
retained_customers = df[df["Churn"] == "No"]
```

```python
# average monthly charges
print("Churned Customers Avg Monthly charges:", np.mean(churned_customers["MonthlyCharges"]))
print("Retained Customers Avg Monthly charges:", np.mean(retained_customers["MonthlyCharges"]))

```

```python
#average tenure
print("Churned Average Tenure:", np.mean(churned_customers["tenure"]))

print("Retained Average Tenure:", np.mean(retained_customers["tenure"]))
```

```python
# Monthly Revenue at Risk
revenue_at_risk = np.sum(churned_customers["MonthlyCharges"])
print("Monthly Revenue at Risk:", revenue_at_risk)
```

```python
# Retained Customers Revenue
retained_revenue = np.sum(retained_customers["MonthlyCharges"])
print("Retained Customers Revenue:", retained_revenue)
```

```python
#avg revenue per customer
avg_revenue = np.mean(df["MonthlyCharges"])
print("Average Revenue per Customer:", avg_revenue)
```

```python
#Correlation Analysis
```

```python
# Tenure and Monthly Charges
corr1 = np.corrcoef(
    df["tenure"],
    df["MonthlyCharges"]
)

print("Tenure vs Monthly Charges:", corr1)


# Tenure and Churn
corr2 = np.corrcoef(
    df["tenure"],
    df["ChurnFlag"]
)

print("Tenure vs Churn:", corr2)
```

```python
# actual values
corr1 = np.corrcoef(df["tenure"], df["MonthlyCharges"])[0, 1]
corr2 = np.corrcoef(df["tenure"], df["ChurnFlag"])[0, 1]

print("Tenure vs Monthly Charges:", corr1)
print("Tenure vs Churn:", corr2)
```

```python
# contract-wise churn correlation
revenue_risk = df[df["Churn"] == "Yes"].groupby("Contract").agg(
    Churned_Customers=("customerID", "count"),
    Revenue_At_Risk=("MonthlyCharges", "sum")
)

print(revenue_risk)
```
